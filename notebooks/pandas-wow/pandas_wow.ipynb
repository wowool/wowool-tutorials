# Wow to pandas app

Wow annotation engine produces a very rich representation of entities. For instance, for a Person, we get the gender, first and family name and sometimes the profession. For a Company, the sector and the country. We can also get the sentence where the entity ocurred as well as the document id. At document level we have topics and themes and, if available, metadata such as date, author, author. 

All this information gets automatically transformed with the pandas.app into a pandas dataframe.

First let's import the necessary libraries. 

```python
from wowool.sdk import Pipeline
from wowool.document import Document
import pandas as pd
from pathlib import Path
```

Let's create a wowool pipeline. There are some defaults already set, but in this case, we want to select the annotations and attributes that interest us.

```python
from wowool.sdk import Pipeline
import pandas as pd

text = """John Doe is a software engineer at Google in USA. He has been working on various projects related to machine learning and artificial intelligence. In his free time, he enjoys hiking and playing the guitar."""

pipeline_description = [
            "english", 
            "entity", 
            "business-news",
            "metadata",
            "semantic-theme", 
            {
                "name": "themes.app",
                "options": {"count": 2}
            },
            {
                "name": "topics.app",
                "options": {"count": 2}
            },
            {
                "name": "dataframe.app",
                "options": {
                    "annotations": ["Person", "Organization", "Company", "Country","CompanyOperation","Product","RawMaterial"],
                    "attributes": ["uri","canonical","country","sector","theme","themes","sent_id"],
                    "metadata": ["date"]
                }
            }
        ]

pipeline = Pipeline(pipeline_description)

input_path = Path("data").expanduser()

dfs = []

for ip in Document.glob(input_path, "*.txt"):

    # Process the input text using the create pipeline
    doc = pipeline(ip.data)
    df = pd.DataFrame(doc.dataframe.rows)
        # Add it to the collection
    print(df)
    exit(-1)
    dfs.append(df)

# Combine all documents into one DataFrame
final_df = pd.concat(dfs, ignore_index=True)
mask = final_df["uri"] == "CompanyOperation"
print(final_df[mask])
```
